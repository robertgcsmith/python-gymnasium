## Part 1
### Find x

We want to find a positive real number $x$ satisfying:
$$x^{x^{x^{2017}}} = 2017$$

Let us verify that $x = 2017^{\frac{1}{2017}}$ (or $\sqrt[2017]{2017}$) is a solution. We substitute $x = 2017^{\frac{1}{2017}}$ into the tower and evaluate from the top down:

$$
\begin{aligned}
\text{Top layer: } \quad x^{2017} &= \left(2017^{\frac{1}{2017}}\right)^{2017} = 2017 \\[1em]
\text{Middle layer: } \quad x^{\left(x^{2017}\right)} &= x^{2017} = 2017 \\[1em]
\text{Full tower: } \quad x^{\left(x^{x^{2017}}\right)} &= x^{2017} = 2017
\end{aligned}
$$

Because the powers cascade and cancel perfectly from the top down, $x = \sqrt[2017]{2017}$ is verified as a valid positive solution.

### Uniqueness

To check whether $x = \sqrt[2017]{2017}$ is a unique solution, let the function $f(x)$ for all positive real numbers $x > 0$:
$$f(x) = x^{x^{x^{2017}}}.$$

The behaviour of any exponential expression $x^{k}$ undergoes a dramatic shift at the boundary of $1$. This can be observed through the derivativee of a basic exponential function $f(k) = x^k$ with respect to the exponent $k$ 

$$\frac{d}{dk}(x^{k})=x^{k}\cdot \ln (x).$$

* The behaviour of this derivative depends entirely on the sign of $\ln(x)$: If $x > 1$, then $\ln(x) > 0 \rightarrow$ the derivative is positive (the function grows).
* If $x = 1$, then $\ln(1) = 0 \rightarrow $ the derivative is zero (the function is perfectly flat).
* If $0 < x < 1$, then $\ln(x) < 0 \rightarrow $ the derivative is negative (the function shrinks). 

To show $f(x) = 2017$ has exactly one solution, we split domains and analyse behaviour in each respective interval:

##### Case 1: $0 < x \le 1$
* If $x = 1$, then $f(1) = 1^1 = 1$.
* If $0 < x < 1$, raising a number between 0 and 1 to any positive power results in another number less than 1. Therefore, $f(x) < 1$.
* Since the function shrinks toward zero, and $2017 > 1$, no solution can exist in the interval $(0, 1]$.

#### Case 2: $x > 1$
For $x > 1$, both the base and the exponent vary simultaneously. To prove that $f(x) = x^{x^{x^{2017}}}$ is strictly increasing, we can use the fundamental property of inequalities for real numbers greater than $1$: 
If $1 < x_1 < x_2$ and $0 < y_1 < y_2$, then $x_1^{y_1} < x_2^{y_2}$.

Doing so, we can build the tower layers inductively for any two values $1 < x_1 < x_2$:

1. **Layer 1 ($x^{2017}$):** Since $x > 1$, the function $g(x) = x^{2017}$ is strictly increasing. Therefore:
   $$x_1^{2017} < x_2^{2017}$$

2. **Layer 2 ($x^{x^{2017}}$):** Consider the expression $x_1^{x_1^{2017}}$. Because the base increases ($x_1 < x_2$) and the exponent increases ($x_1^{2017} < x_2^{2017}$), and both are strictly greater than $1$, the value must strictly grow:
   $$x_1^{x_1^{2017}} < x_2^{x_1^{2017}} < x_2^{x_2^{2017}} \implies x_1^{x_1^{2017}} < x_2^{x_2^{2017}}$$

3. **Full Tower ($f(x)$):** Applying this exact same logic to the final layer, where both the base $x$ and the entire upper exponent tower are strictly positive and strictly increasing for $x > 1$:
   $$f(x_1) < f(x_2)$$

Because $f(x)$ is a composition of functions where both base and exponent grow strictly over the domain $x > 1$, the overall function $f(x)$ is strictly increasing. It passes the horizontal line test and crosses the value $2017$ exactly once.


### Conclusion
Since $x = \sqrt[2017]{2017} \approx 1.00378 > 1$, it lies firmly in the strictly increasing region; therefore, it is a unique positive real solution.


```python
"""part 2"""
import math

def solve_original() -> float:
    """
    Return the positive real solution to
    x ** (x ** (x ** 2017)) = 2017.

    Derive the solution mathematically.
    Do not use numerical root-finding.
    """
    return 2017**(1/2017)

def evaluate_original(x: float) -> float:
    """
    Evaluate the original power tower,
    respecting exponentiation order.
    """
    return x ** (x ** (x ** 2017))

# verification
x = solve_original()

assert x > 0
assert math.isclose(
    evaluate_original(x),
    2017.0,
    rel_tol=1e-8
)

print(f"Computed x: {x}")
print(f"Evaluated tower: {evaluate_original(x)}")
```

## Part 3
The mathematical insight from Part 1 generalises for any integer $n \geq 2$. For the equation $x^{x^{x^n}} = n$, the unique positive real solution is:
$$x = n^{\frac{1}{n}} = \sqrt[n]{n}$$

In exact real arithmetic, evaluating the power tower from the top down allows the exponents to collapse perfectly layer by layer back to $n$, preventing the numbers from blowing up or overflowing.

However, as demonstrated below, this identity does not hold true in finite precision floating point arithmetic, where compounding rounding errors dramatically alter the behaviour of the tower for large values of $n$.


```python
"""generalised implementation"""

import math

def solve_tower(n: int) -> float:
    """
    Return the unique positive real solution to
    x ** (x ** (x ** n)) = n for any integer n >= 2.
    """
    return n ** (1 / n)

def evaluate_tower(x: float, n: int) -> float:
    """
    Evaluate the generalised power tower from top to bottom.
    """
    return x ** (x ** (x ** n))

# Test cases
test_cases = [2, 3, 10, 100, 2017, 10**8, 10**15, 10**18]

print(f"{'n':<10} | {'Computed x':<20} | {'Evaluated Tower':<25} | {'Observed Outcome / Error':<20}")
print("-" * 88)

for n in test_cases:
    try:
        x = solve_tower(n)
        result = evaluate_tower(x, n)
        abs_error =  abs(result - n)
    
        print(f"{n:<10} | {x:<20.14f} | {result:<25.14f} | Err: {abs_error:.5e}")
    except OverflowError:
        print(f"{n:<10} | {x:<20.14f} | {'OVERFLOW':<25} | Raised OverflowError")

print("\n")
print("Numerical analysis + properties:")
print("Floating point failures are highly non-linear and do not always manifest as errors:")
print("1. For n = 10^8:  Returns a finite result but carries an absolute error of ~197.42.")
print("2. For n = 10^15: Compounding rounding errors cause the outermost exponentiation to overflow.")
print("3. For n = 10^18: The value 1.000000000000000055... rounds exactly to 1.0 in float64 storage.")
print("The tower collapses silently to 1.0 without throwing any exception.")
```

To rigorously prove that $f(x) = x^{x^{x^{2017}}}$ is strictly increasing for $x > 1$ while accounting for both base and exponent varying, we can decompose the tower using natural logarithms.
Let us define the nested layers of the power tower as functions for $x > 1$:
1. $h_1(x) = x^{2017}$
2. $h_2(x) = x^{h_1(x)} = e^{h_1(x) \ln x}$
3. $f(x) = x^{h_2(x)} = e^{h_2(x) \ln x}$ \
We evaluate monotonicity from the bottom layer up using a foundational calculus rule: the product of two strictly positive and strictly increasing functions is itself strictly increasing.
- Layer 1: For $x > 1$, $h_1(x) = x^{2017}$ is strictly positive and its derivative $2017x^{2016} > 0$, so $h_1(x)$ is strictly increasing.
- Layer 2: Consider the exponent of the base-$e$ representation, $g_2(x) = h_1(x) \ln x$. \
    * For $x > 1$, both $h_1(x)$ and $\ln x$ are strictly positive and strictly increasing. \
    * Therefore, their product $g_2(x)$ is strictly increasing. \
    * Since the natural exponential function $e^{t}$ is strictly increasing, $h_2(x) = e^{g_2(x)}$ must be strictly increasing. (Note that since $g_2(x) > 0$, $h_2(x) > 1$). \
- Layer 3 (Full Tower): Consider the final exponent, $g_3(x) = h_2(x) \ln x$. \
    * Since $h_2(x) > 1$ and is strictly increasing, and $\ln x > 0$ and is strictly increasing, their product $g_3(x)$ is strictly increasing. \
    *  Consequently, $f(x) = e^{g_3(x)}$ is strictly increasing for all $x > 1$. \
By demonstrating that the product of the shifting base and shifting exponent preserves growth at every single layer, the proof is complete without executing a messy chain-rule derivative.

ii)
Why the assertion passes:
The math.isclose(result, n, rel_tol=1e-7) function checks relative error, which is defined as:
$\text{Relative\ Error}=\frac{|{}\text{result}-n|{}}{|{}n|{}}$
When $n\$ scales to a massive number like $10^{8}$, an absolute error of $\approx 197.42$ yields a relative error of
$\frac{197.42}{10^{8}}=1.97424 \times 10^{-6}.$
Because $1.97424 \times 10^{-6} > 1 \times 10^{-7}$, the math.isclose assertion with a tolerance of 1e-7 will fail.

Relative error is the superior metric when analyzing scale-invariant systems or data spanning multiple orders of magnitude. However, it is important to distinguish relative error (a descriptive measure of an approximation's objective accuracy) from relative tolerance (a prescriptive pass/fail threshold set by the application design). Passing a tolerance check does not imply a value is universally accurate; it only means it is acceptable under that specific constraint.

Moreover, the mathematical identity \(x = n^{1/n}\) is exact, representing a precise point on the real number line, which is often an irrational number. Since computers have a finite amount of memory, fractions like 1/2017 or 1/3 cannot be perfectly expressed in base-2 binary; modern binary64 floating-point systems do not simply truncate trailing bits; they use a strict round-to-nearest, ties-to-even convention. Floating-point exponentiation (**) is highly implementation, and platform, dependent. Rather than assuming a specific underlying algorithm (like CORDIC or polynomial Taylor expansions), we treat the operation as an approximate calculation subject to standard rounding limits.

Therefore, the formula possesses perfect algebraic correctness, but the computer can only construct a high-precision digital proxy of that number. The numerical discrepancy is driven by an accumulation of errors across three distinct stages:
- 1. Representing the fraction \(1/n\) in binary.
- 2. Evaluating the base calculation n ** (1 / n).
- 3. Propagating those initial precision losses exponentially upward through successive layers of the power tower.
