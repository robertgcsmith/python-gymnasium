# Notes: time and space complexity analysis

1. Time complexity breakdown
   - get(key): Average/amortised O(1). Python dictionary operations execute keys 
     on average in constant time. Repositioning an existing active node to the MRU 
     front requires executing a combined total of six structural link pointer 
     modifications (2 adjustments during _remove and 4 adjustments during _add). 
     This structural transaction routine scales entirely independent of cache size.
   - put(key, value): Average/amortised O(1). Modifying active dictionary nodes 
     recycles the specialized Node structure in place. Brand-new entries use 
     constant-time dictionary tracking additions alongside fixed sentinel node 
     splicing chains.
   - __len__() / __contains__(): Average O(1). Evaluates instantly from the 
     internal collection metrics.
   - __repr__(): O(N). Demands explicit sequential loop traversal tracking elements 
     from the dummy head to tail node to construct output literals.

2. Space complexity breakdown
  - Overall Footprint: O(C), where C is the maximum allowed capacity limit. 
     The memory overhead scales linearly with respect to the limit. The object handles 
     at most C data instances and exactly 2 base structural node entities simultaneously.
